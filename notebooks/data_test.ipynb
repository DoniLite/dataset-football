### Random data generation for testing all working in the database

```python
from datetime import datetime
import random
from sqlalchemy.orm import Session
from config.db import session
from config.schemas import Match, Team, Player

def create_test_data(session: Session):
    # Création des équipes et joueurs
    teams = []
    team_names = [
        "Paris Saint-Germain", "Manchester City", "Real Madrid", "Bayern Munich",
        "Liverpool", "Barcelona", "Juventus", "Ajax", "Porto", "Inter Milan",
        "Chelsea", "Borussia Dortmund", "Atletico Madrid", "AC Milan", "Benfica",
        "Arsenal", "Napoli", "RB Leipzig", "Sevilla", "Lyon"
    ]
    
    # Créer les équipes et leurs joueurs
    for team_name in team_names:
        team = Team(name=team_name)
        
        # Créer des joueurs pour chaque équipe
        for i in range(11):  # 11 joueurs par équipe
            player = Player(
                full_name=f"Player {i+1} of {team_name}",
                team=team,
                player_number=(i + random.randint(0, 26))
            )
            team.players.append(player)
        
        teams.append(team)
        session.add(team)
    
    # Créer 10 matchs de championnat
    for i in range(10):
        # Sélectionner deux équipes différentes au hasard
        team_a, team_b = random.sample(teams, 2)
        
        # Générer des statistiques aléatoires
        team_a_goal = random.randint(0, 5)
        team_b_goal = random.randint(0, 5)
        
        match = Match(
            # Informations temporelles
            date_text=datetime.now().strftime("%Y-%m-%d"),
            unix_timestamp=str(int(datetime.now().timestamp())),
            day=f"Journée {i+1}",
            
            # Identifiants
            match_id=random.randint(1, 5000),
            coupe_id=random.randint(1, 5000),
            
            # Équipe A
            team_a_logo_url=f"https://example.com/{team_a.name.lower().replace(' ', '_')}.png",
            team_a_name=team_a.name,
            team_a_short=team_a.name[:3].upper(),
            team_a_goal=str(team_a_goal),
            team_a_shots_on_target=str(random.randint(team_a_goal, team_a_goal + 5)),
            team_a_possession=str(random.randint(35, 65)),
            team_a_passes=str(random.randint(300, 700)),
            team_a_pass_accuracy=str(random.randint(70, 90)),
            team_a_fouls=str(random.randint(5, 15)),
            team_a_yellow_cards=str(random.randint(0, 3)),
            team_a_red_cards=str(random.randint(0, 1)),
            team_a_offsides=str(random.randint(0, 5)),
            team_a_corners=str(random.randint(2, 10)),
            team_a_goal_times=",".join([str(random.randint(1, 90)) for _ in range(team_a_goal)]),
            
            # Équipe B
            team_b_logo_url=f"https://example.com/{team_b.name.lower().replace(' ', '_')}.png",
            team_b_name=team_b.name,
            team_b_short=team_b.name[:3].upper(),
            team_b_goal=str(team_b_goal),
            team_b_shots_on_target=str(random.randint(team_b_goal, team_b_goal + 5)),
            team_b_possession=str(100 - int(random.randint(35, 65))),
            team_b_passes=str(random.randint(300, 700)),
            team_b_pass_accuracy=str(random.randint(70, 90)),
            team_b_fouls=str(random.randint(5, 15)),
            team_b_yellow_cards=str(random.randint(0, 3)),
            team_b_red_cards=str(random.randint(0, 1)),
            team_b_offsides=str(random.randint(0, 5)),
            team_b_corners=str(random.randint(2, 10)),
            team_b_goal_times=",".join([str(random.randint(1, 90)) for _ in range(team_b_goal)])
        )
        
        session.add(match)
    
    # Sauvegarder toutes les données
    session.commit()
    
try:
    create_test_data(session)
    print("Données de test créées avec succès!")
            
    # Vérifier les données
    print("\nÉquipes créées:")
    teams = session.query(Team).all()
    for team in teams:
        print(f"- {team.name} ({len(team.players)} joueurs)")
            
    print("\nMatchs créés:")
    matches = session.query(Match).all()
    for match in matches:
        print(f"- {match.team_a_name} {match.team_a_goal}-{match.team_b_goal} {match.team_b_name}")
                
except Exception as e:
    print(f"Erreur lors de la création des données : {e}")
    session.rollback()
```
